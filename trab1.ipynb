```python
# Trabalho 1
# Busca Incremental

import numpy as np

def func(x):
    return ((np.exp(x)) - 4) + (7/2) * np.cos(2 * x)

eixo_x = np.linspace(0, 3, 101) #vetor de intervalos

lower = func(eixo_x[0])

for i in range(1, len(eixo_x)):
    upper = func(eixo_x[i])

    if(upper * lower <= 0):
        print(f"Possível raiz entre {eixo_x[i - 1]:.4f} e {eixo_x[i]:.4f}")
    lower = upper 
```

### TRABALHO 1B | MÉTODO DA BISSECÇÃO

```python
# Trabalho 1
# Bissecao

import numpy as np

def erro_aprox(x, x_prox):

  epsilon = abs( ((x_prox - x) / x_prox) * 100 )
  return epsilon

def func(x):
    return ((np.exp(x)) - 4) + (7/2) * np.cos(2 * x)

alvo_loop = 100 #epsilon
eixo_x = np.linspace(0, 3, 101) #vetor de intervalos
lower = eixo_x[0]
raiz_atual = 0


for i in range(1, len(eixo_x)):
    upper = eixo_x[i]

    if(func(upper) * func(lower) <= 0): #encontrou raiz, executa biseccao
        while (alvo_loop > 0.01):
            xm = (upper + lower) / 2
            raiz_prox = xm
            
            if(func(upper) * func(xm) < 0): #raiz esta a a direita do ponto medio 
                lower = xm
            
            else:                           #raiz esta a esquerda do ponto medio
                upper = xm

            alvo_loop = erro_aprox(raiz_atual, raiz_prox)
            raiz_atual = raiz_prox


        break

    lower = upper         



print(f"RAIZ FINAL: {xm:.4f}")
    
```

### TRABALHO 1C | MÉTODO DA FALSA POSIÇÃO

```python
# Trabalho 1
# Falsa posicao

import numpy as np

def erro_aprox(x, x_prox):

  epsilon = abs( ((x_prox - x) / x_prox) * 100 )
  return epsilon

def func(x):
    return ((np.exp(x)) - 4) + (7/2) * np.cos(2 * x)

alvo_loop = 100 #epsilon
eixo_x = np.linspace(0, 3, 101) #vetor de intervalos
lower = eixo_x[0]
raiz_atual = 0


for i in range(1, len(eixo_x)):
    upper = eixo_x[i]

    if(func(upper) * func(lower) <= 0): #encontrou raiz, executa falsa posicao
        while (alvo_loop > 0.01):
            xr = ( lower * func(upper) - upper * func(lower) ) / ( func(upper) - func(lower) )
            raiz_prox = xr
            
            if(func(upper) * func(xr) < 0): #raiz esta a direita dda falsa posicao
                lower = xr
            
            else:                           #raiz esta a esquerda da falsa posicao
                upper = xr

            alvo_loop = erro_aprox(raiz_atual, raiz_prox)
            raiz_atual = raiz_prox


        break

    lower = upper         



print(f"RAIZ FINAL: {xr:.4f}")
    
```
