### Otimização | Razão Áurea

```python
# Trabalho 2A
# Otimizacao por razao aurea

phi = 1.618033988749895

def func(x):
    return (-3 * x ** 6) + (3 * x ** 4) - (2 * x) + 5

def erro(xl, xu, xopt):
    return (2 - phi) * abs((xu - xl) / xopt) * 100

def dis(xl, xu):
    return (phi - 1) * (xu - xl)


# loop principal
xl = -3
xu = 2
d = dis(xl, xu)
x1 = xl + d
x2 = xu - d

f1 = func(x1)
f2 = func(x2)

# melhor xopt
if f1 > f2:
    xopt = x1
else:
    xopt = x2

while erro(xl, xu, xopt) > 1e-8:

    if f1 > f2: # ponto mais altoa direita
        xl = x2
        x2 = x1
        d = dis(xl, xu)
        x1 = xl + d
        f2 = f1
        f1 = func(x1)

    else:                   # ponto mais alto a esquerda
        xu = x1
        x1 = x2
        d = dis(xl, xu)
        x2 = xu - d
        f1 = f2
        f2 = func(x2)

    # atualizar xopt
    if f1 > f2:
        xopt = x1
    else:
        xopt = x2



print(f"x no valor máximo: {xopt:.4f}")
print(f"f(x): {func(xopt):.4f}")

```

### Interpolação quadrátrica

```python
# Trabalho 2B
# interpolacao quadratica


def func(x):
    return (-3 * x ** 6) + (3 * x ** 4) - (2 * x) + 5

def erro(x_atual, x_ant):
    return abs(x_atual - x_ant) / abs(x_ant)

def interpolacao(x1, x2, x3):
    return x2 + (-1/2 * ( (x2 - x1) ** 2 * (func(x2) - func(x3)) - (x2 - x3) ** 2 * (func(x2) - func(x1))) / ( (x2 - x1) * (func(x2) - func(x3)) - (x2 - x3) * (func(x2) - func(x1))) )

# intervalo [-3, 2]
x1 = -3
x3 = 2
x2 = (x1 + x3) / 2

x4 = interpolacao(x1, x2, x3)
erro_estimado = erro(x4, x2)

while erro_estimado > 1e-8:
    
    if x4 > x2:
        if func(x4) > func(x2):
            x1 = x2
            x2 = x4
        else:
            x3 = x4
    else:
        if func(x4) > func(x2):
            x3 = x2
            x2 = x4
        else:
            x1 = x4
        
    x_ant = x2
    x4 = interpolacao(x1, x2, x3)
    erro_estimado = erro(x4, x_ant)


print(f"x no valor máximo: {x4:.4f}")
print(f"f(x): {func(x4):.4f}")

```
