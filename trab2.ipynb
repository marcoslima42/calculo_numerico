### Otimização | Razão Áurea

```python
# Trabalho 2A
# Otimizacao por razao aurea

phi = 1.618033988749895

def func(x):
    return (-3 * x ** 6) + (3 * x ** 4) - (2 * x) + 5

def erro(xl, xu, xopt):
    return (2 - phi) * abs((xu - xl) / xopt) * 100

def dis(xl, xu):
    return (phi - 1) * (xu - xl)


# loop principal
xl = -3
xu = 2
d = dis(xl, xu)
x1 = xl + d
x2 = xu - d

f1 = func(x1)
f2 = func(x2)

# melhor xopt
if f1 > f2:
    xopt = x1
else:
    xopt = x2

while erro(xl, xu, xopt) > 1e-8:

    if f1 > f2: # ponto mais altoa direita
        xl = x2
        x2 = x1
        d = dis(xl, xu)
        x1 = xl + d
        f2 = f1
        f1 = func(x1)

    else:                   # ponto mais alto a esquerda
        xu = x1
        x1 = x2
        d = dis(xl, xu)
        x2 = xu - d
        f1 = f2
        f2 = func(x2)

    # atualizar xopt
    if f1 > f2:
        xopt = x1
    else:
        xopt = x2



print(f"x no valor máximo: {xopt:.10f}")
print(f"f(x): {func(xopt):.10f}")

```
