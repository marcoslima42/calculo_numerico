```python
# Trabalho 4
# Ponto Fixo Simples

def erro_aprox(x, x_prox):

  epsilon = abs( ((x_prox - x) / x_prox) * 100 )
  return epsilon

import matplotlib.pyplot as plt
import numpy as np
import math

#f(x) = e^-x - x
alvo_loop = 100 #epsilon
raiz_atual_g1 = 0
raiz_prox_g1 = None

while True:

  #erro de g1
  raiz_prox_g1= math.exp(-raiz_atual_g1)
  alvo_loop = erro_aprox(raiz_atual_g1, raiz_prox_g1)
  raiz_atual_g1 = raiz_prox_g1

  #g2 e uma funcao sem imagem em zero logo g1 sempre sera raiz-math.log(0)
  #além disso a derivada de g2 quando x diferente de 1 sempre divergira

  if(alvo_loop < 0.01):
    break

print('RAIZ FINAL: ', raiz_atual_g1)
```
